## More Type Conversions

Sometimes we need to change data from one type to another so we can work with it properly.

### Examples:

````python
# i. Convert "25" to integer
number_str = "25"
number_int = int(number_str)
print(f"'25' as int: {number_int} (Type: {type(number_int)})")

# ii. Convert 15 to string
age = 15
age_str = str(age)
print(f"15 as str: '{age_str}' (Type: {type(age_str)})")

# iii. Convert "7.5" to float
price_str = "7.5"
price_float = float(price_str)
print(f"'7.5' as float: {price_float} (Type: {type(price_float)})")

# iv. Convert 0 to boolean
is_active_num = 0
is_active_bool = bool(is_active_num)
print(f"0 as bool: {is_active_bool} (Type: {type(is_active_bool)})")
````
